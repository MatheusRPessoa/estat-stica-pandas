```python
import pandas as pd
```

```python
dict_metrics = {
    'idade': [15, 18, 25, 25, 40, 55, 58, 60, 80], 
    'altura': [160, 162, 165, 168, 172, 174, 174, 174, 176]
}
```

```python
print(dict_metrics)
```

```python
df_metrics = pd.DataFrame.from_dict(dict_metrics)
df_metrics
```

## Medidas Resumo

### Medidas de posição

```python
# Média
df_metrics["idade"].mean()
```

```python
df_metrics["idade"].median()
```

```python
df_metrics["idade"].mode()
```

```python
df_metrics["altura"].mode()
```

### Medidas de dispersão

```python
# Variância
df_metrics["idade"].var()
```

```python
# Desvio padão
df_metrics["idade"].std()
```

```python
# Coeficiente de variação - IDADE
desvio_padrao = df_metrics["idade"].std()
media = df_metrics["idade"].mean()
desvio_padrao / media * 100
```

```python
# Coeficiente de variação - ALTURA
desvio_padrao = df_metrics["altura"].std()
media = df_metrics["altura"].mean()
desvio_padrao / media * 100
```

### Medidas de forma

```python
# Assimetria = Idade
df_metrics["idade"].skew()
```

```python
# Assimetria = Altura
df_metrics["altura"].skew()
```

```python
# Curtose = Idade
df_metrics["idade"].kurtosis()
```

```python
df_metrics["altura"].kurtosis()
```

```python
# Obter medidas estatisticas
df_metrics["idade"].describe()
```

### Correlação de variáveis

```python
# Correlação do DataFrame
df_metrics.corr(method='pearson')
```

```python
# Correlação entre duas variáveis
df_metrics.idade.corr(df_metrics.altura)
```

```python
# Correlação do Dataframe, segundo Spearman
df_metrics.corr(method='spearman')
```
