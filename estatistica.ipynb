```python
import pandas as pd
```

```python
dict_metrics = {
    'idade': [15, 18, 25, 25, 40, 55, 58, 60, 80], 
    'altura': [160, 162, 165, 168, 172, 174, 174, 174, 176]
}
```

```python
print(dict_metrics)
```

```python
df_metrics = pd.DataFrame.from_dict(dict_metrics)
df_metrics
```

## Medidas Resumo

### Medidas de posição

```python
# Média
df_metrics["idade"].mean()
```

```python
df_metrics["idade"].median()
```

```python
df_metrics["idade"].mode()
```

```python
df_metrics["altura"].mode()
```

### Medidas de dispersão

```python
# Variância
df_metrics["idade"].var()
```

```python
# Desvio padão
df_metrics["idade"].std()
```

```python
# Coeficiente de variação - IDADE
desvio_padrao = df_metrics["idade"].std()
media = df_metrics["idade"].mean()
desvio_padrao / media * 100
```

```python
# Coeficiente de variação - ALTURA
desvio_padrao = df_metrics["altura"].std()
media = df_metrics["altura"].mean()
desvio_padrao / media * 100
```

### Medidas de forma

```python
# Assimetria = Idade
df_metrics["idade"].skew()
```

```python
# Assimetria = Altura
df_metrics["altura"].skew()
```

```python
# Curtose = Idade
df_metrics["idade"].kurtosis()
```

```python
df_metrics["altura"].kurtosis()
```

```python
# Obter medidas estatisticas
df_metrics["idade"].describe()
```

### Correlação de variáveis

```python
# Correlação do DataFrame
df_metrics.corr(method='pearson')
```

```python
# Correlação entre duas variáveis
df_metrics.idade.corr(df_metrics.altura)
```

```python
# Correlação do Dataframe, segundo Spearman
df_metrics.corr(method='spearman')
```

## Representações Gráficas

```python
# Histograma - Idade
df_metrics.idade.hist()
```

### Criar DataFrame de Vendas

```python
dict_vendas = {
    'categoria': ['masculino', 'feminino', 'infantil', 'casa'],
    'valor': [400000, 600000, 250000, 580000],
    'quantidade': [3000, 5000, 1500, 2500]
}
```

```python
df_vendas = pd.DataFrame.from_dict(dict_vendas)
```

```python
df_vendas
```

```python
# Gráfico de barras vertical - valor
df_vendas.plot.bar(x='categoria',y='valor')
```

```python
# Gráfico de barras vertical - valor (ordenado)
df_vendas.sort_values('valor', ascending=False).plot.bar(x='categoria', y='valor')
```

```python
# Gráfico de barras horizontal - Quantidade
df_vendas.sort_values('quantidade').plot.barh(x='categoria', y='quantidade')
```
